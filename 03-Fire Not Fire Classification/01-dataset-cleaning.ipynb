```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
dataset = pd.read_csv('Raw-Algerian_forest_fires_dataset_UPDATE.csv', header = 1)
dataset.head(3)
```

# Dataset Cleaning

- Dataset is in two parts (Region)
- Merge them with a category 0 & 1 as per Region: (:122) as 0,  (122:) as 1 and Merge


```python
dataset[dataset.isnull().any(axis=1)]
```

```python
# Merging Data Set
df = dataset
df.loc[:122, "Region"] = 0
df.loc[122:, "Region"] = 1

df[["Region"]] = df[["Region"]].astype(int)
```

```python
# Removing Nan and Redundant row
df.isnull().sum()
df = df.dropna().reset_index(drop=True)

df.iloc[122]
df = df.drop(122).reset_index(drop=True)
```

```python
# Cleaing Column Name

df.columns = df.columns.str.strip()
df.columns
```

```python
# Converting DataType

df.info()
df['Temperature'].unique()

df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']] = df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
objects = [features for features in df.columns if df[features].dtypes=='O']

for i in objects:
    if i != 'Classes':
        df[i] = df[i].astype(float)

df.info()
```

```python
df.to_csv('Cleaned-Algerian_forest_fires_dataset.csv', index=False)
```
